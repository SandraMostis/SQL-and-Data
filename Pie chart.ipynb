```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

Activity= ['Meeting', 'Presentation', 'Lunch', 'Reading']
Minutes = [60, 30, 40, 30]
explode2 =(0, 0, 0, 0.1)
ColorList =('orange', 'cyan', 'green', 'grey')


fig1, ax1 = plt.subplots(figsize =(5, 7))
ax1.pie(Minutes, colors=ColorList, explode = explode2, labels=Activity, autopct = '%1.f%%')

plt.legend(loc='upper right')
plt.show()
```
