```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DF2 = pd.DataFrame(np.random.rand(20, 5), 
                   columns=['PP', 'QQ', 'RR', 'SS', 'TT'])
DF2.plot.barh(stacked=True, color=['red', 'green', 'blue', 'cyan', 'orange'])
plt.grid(True)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Horizontal Stacked Bar Plot')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

Data = {'Month':['Jan','Feb','Mar','Apr'],
'Sale Value':[8000, 5000, 6000, 4000],
'Expenditure Value':[10000, 4000, 6000, 6000]}

DataFrame1 = pd.DataFrame(Data)
print(DataFrame1)

DataFrame1.plot.barh(stacked=True, color=['cyan', 'green'])

plt.xlabel('Value')
plt.ylabel('Months')
plt.title('L2 Try It Yourself Exercise - Horizontal Bar Plot')
plt.show()
```
