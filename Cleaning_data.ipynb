```python
import pandas as pd
from pandas import DataFrame
import numpy as np

DC = {'Age' : ['50', '30', '19', '25', '40', '55', '35'],
      'Income' : ['55000', '33000', '', '26000', '', '4000', '35000'],
      'Occupation' : ['Consultant', 'Builder', 'Waister', 'Engineer', 'Lecturer', '', ''],
      'Telephone' : ['011111111', '022222222', '01', '044444444', '03333333', '', '077777777'],
      'CreditRating' : ['1', '2', '5', '2', '3', '','4'],
      }
DataFrame1 = DataFrame(DC)
DataFrame1.loc[[2, 4], 'Income'] = np.nan
DataFrame1.loc[[5, 6], 'Occupation'] = np.nan
DataFrame1.loc[5,'Telephone'] = np.nan
DataFrame1.loc[5,'CreditRating'] = np.nan

DataFrame1.loc[7] = ['30', '3000', 'Lec', '088888888', '5'] 
DataFrame1.loc[8] = ['35', '4000', 'Lec', '099999999', '7']

DataFrame1['Occupation'] = DataFrame1['Occupation'].replace('Lec', 'Lecturer', inplace=True)
DataFrame1['Telephone'] = DataFrame1['Telephone'].replace('01', 'NaN', inplace=True)

print(DataFrame1)
```
