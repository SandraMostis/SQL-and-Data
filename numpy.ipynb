```python
import matplotlib.pyplot as plt
import numpy as np
x = np.arange(10)
b = np.arange(10)
plt.plot(b, b**2, label ='b**2') 
plt.plot(b, b**3, label = 'b**3')
plt.plot(b, b**4,label = 'b**4')
plt.legend()

plt.xlabel('x = np.arange(10)')
plt.ylabel('y=x')
plt.title("Simple Plot Demo")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
Data =[0, 11, 12, 13, 14, 15, 16, 17]
Index =[0, 1, 2, 3, 4, 5, 6, 7]
Data2 = [0, 22,24,26,28,30,32,34]
plt.plot(Data, Index, 'o--', label='Data')
plt.plot(Data2, Index, '^--', label='Data2') 
plt.legend()

plt.xlabel('Value')
plt.ylabel('Index')
plt.title("Try it yourself")
plt.savefig('test.png')
```
