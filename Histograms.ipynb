```python
#customised colours 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
x = [60, 90, 67, 76, 61, 78, 67, 75, 64, 66, 75, 
     69, 87, 73, 56, 92, 77, 73, 82, 74, 75, 72, 
     95, 69, 60, 74, 67, 56, 63]
plt.hist(x, bins=[40, 50, 60, 70,80, 90, 100], color='yellow', edgecolor='black', linewidth=2)
plt.xlabel('s= results')
plt.ylabel('y= total students')
plt.show()
```
