```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

a=np.array([48, 49, 54, 78, 57, 69, 52, 77, 90])
b=np.array([150, 155, 160, 148, 150, 162, 178, 180, 175])
c=np.array([50, 52, 56, 75, 60, 65, 60, 73, 82])
d=np.array([150, 155, 160, 148, 150, 162, 178, 180, 175])

# data about people's weight and height
plt.scatter(a,b,label ='Original')
plt.xlabel('A = Weight in KG')
plt.ylabel('B = Height in CM')
plt.legend(loc = 'center right')

# data about people's weight and height after 2 years
plt.scatter(c, d, label ='After 2 years')
plt.xlabel('A = Weight in KG')
plt.ylabel('B = Height in CM')
plt.legend(loc = 'center right')

plt.show()
```
